```python
import numpy as np

def torus_inner_radius(r: float, p: float, z: int):
    """
    :param r: radius from center of tube to center of torus ring
    :param p: radius of tube
    :param z: height where z=0 is center of torus ring
    """
    if z < -p or z > p:
        return r
    return r - np.sqrt(p ** 2 - z ** 2)

def get_anchor_coordinate_list(cylinder_layers: int, fg_per_layer: int, outer_r: float, half_slab_w: float):
    """for anchoring fgs on torus."""
    coordinates = []
    for i in range(cylinder_layers):
        theta1 = np.pi * i / (cylinder_layers - 1) - (np.pi / 2)
        z = half_slab_w * np.sin(theta1)
        inner_r = torus_inner_radius(outer_r, half_slab_w, z)
        for j in range(int(fg_per_layer)):
            theta2 = 2.0 * np.pi * j / fg_per_layer
            x = inner_r * np.cos(theta2)
            y = inner_r * np.sin(theta2)
            coordinates.append([x, y, z])
    return coordinates

print(get_anchor_coordinate_list(4, 8, 185, 75))
print(get_anchor_coordinate_list(4, 8, 220, 75))
print(get_anchor_coordinate_list(4, 8, 255, 75))
print(get_anchor_coordinate_list(4, 8, 295, 75))
```
