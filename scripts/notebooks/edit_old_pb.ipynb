```python
def edit_pb(old_path, new_path):
    # Read old file
    with open(old_path, "rb") as f:
        output = Output()
        output.ParseFromString(f.read())
        
    # Make changes
    assignment = output.assignment
    assignment.is_multiple_hdf5s = False # was True
    assignment.is_xyz_hist_stats = True # was True
    assignment.full_output_statistics_interval_factor = 1 # was 100, when 1, output every output_statistics_interval_ns
    assignment.simulation_time_ns = 100 # was 100
    
    ns_100_in_frames = 143166
    
    assignment.dump_interval_ns = 10 # was 1
    assignment.dump_interval_frames = ns_100_in_frames / 10
    
    assignment.statistics_interval_ns = 100 # was 0.1
    assignment.statistics_interval_frames = ns_100_in_frames
    
    assignment.output_statistics_interval_ns = 100 # was 1
    assignment.output_statistics_interval_frames = ns_100_in_frames
    
    # Write new file
    with open(new_path, "wb") as nf:
        nf.write(output.SerializeToString())
        print(output.assignment)
```
